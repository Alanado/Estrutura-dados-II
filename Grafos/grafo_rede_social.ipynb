```python
rede = {
    "Ana": ["Bruno", "Carla"],
    "Bruno": ["Ana", "Carla", "Diego"],
    "Carla": ["Ana", "Bruno", "Elisa"],
    "Diego": ["Bruno", "Elisa", "Felipe"],
    "Elisa": ["Carla", "Diego", "Felipe"],
    "Felipe": ["Diego", "Elisa"]
}

# Quantidade de usuarios
n = len(rede)

# Calcule o grau de cada usuario
for usuario, amigos in rede.items():
    print(usuario, "tem", len(amigos), "conexoes")

# A soma dos graus conta cada amizade duas vezes
soma_graus = sum(len(amigos) for amigos in rede.values())
arestas = soma_graus // 2

print("Total de usuarios:", n)
print("Total de amizades:", arestas)

# Densidade do grafo
densidade = (2 * arestas) / (n * (n - 1))
print(f"Densidade: {densidade:.2%}")


# Rede ampliada

for novo in ["Gabriela", "Hugo", "Isabela", "João"]:
    rede[novo] = []

novas = [("Felipe", "Gabriela"), ("Gabriela", "Hugo"), ("Hugo", "Isabela"), ("Isabela", "João"), ("Elisa", "João")]

for vertice_a, vertice_b in novas:
  rede[vertice_a].append(vertice_b)
  rede[vertice_b].append(vertice_a)

print("\nCALCULADOS DA REDE AMPLIADA: ")
print()
# Rede com usuários adicionados
n = len(rede)
print(f"Nova rede: {n}")


# Grau de cada usuário na rede ampliada
for usuario, amigos in rede.items():
    print(usuario, "tem", len(amigos), "conexoes")


# Soma de graus na rede amplicada
soma_graus = sum(len(amigos) for amigos in rede.values())
arestas = soma_graus // 2
print("Total de usuarios:", n)
print("Total de amizades:", arestas)

# Densidade do grafo na rede ampliada
densidade = (2 * arestas) / (n * (n - 1))
print(f"Densidade: {densidade:.2%}")
```
